```python
import os
from google import genai
from pypdf import PdfReader
from dotenv import load_dotenv
```

```python
load_dotenv(dotenv_path='../../../.env')
```

```python
def lePDF(pdf_path):
    
    reader = PdfReader(pdf_path)

    # Total de páginas
    total_paginas = len(reader.pages)

    # Extração do texto de cada página.
    texto_completo = ""
    for i, page in enumerate(reader.pages):
        texto_pagina = page.extract_text()
        texto_completo += texto_pagina + "\n"

    return texto_completo
```

```python
texto_seguro = lePDF('Regulamento de Seguro Fictício.pdf')
```

```python
def criaPrompt(text, user_question):

    prompt = f"""
    Você é uma analista de seguros.
    Sua função é responder a pergunta do usuário baseado no regulamento que contém as regras do seguro. 
    Não invente respostas. 
    Caso não consiga responder a resposta do usuário baseado no texto, retorne "Não foi possível responder a pergunta porque está fora do contexto do regulamento do seguro."

    Regulamento do seguro: [[{text}]]

    Pergunta do usuário: [[{user_question}]]
    """

    return prompt
```

```python
user_question = 'O que acontece se eu atrasar o pagamento do seguro? Qual é a regra que diz sobre atraso de pagamento?'

prompt = criaPrompt(text=texto_seguro, user_question=user_question)

# Inicializa o cliente padrão (ele busca automaticamente a variável GEMINI_API_KEY)
client = genai.Client(api_key=os.environ['GEMINI_API_KEY'])

response = client.models.generate_content(
    model="gemini-3.1-flash-lite",
    # model="gemini-3.8-flash",
    contents=prompt,
)

print(response.text)
```
