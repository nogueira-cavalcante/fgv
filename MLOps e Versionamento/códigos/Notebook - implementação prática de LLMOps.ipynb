```python
import os
from google import genai
from dotenv import load_dotenv
```

```python
load_dotenv(dotenv_path='../../.env')
```

```python
# Inicializa o cliente padrão (ele busca automaticamente a variável GEMINI_API_KEY)
client = genai.Client(api_key=os.environ['GEMINI_API_KEY'])
```

```python
response = client.models.generate_content(
    # model="gemini-3.5-flash",
    model="gemini-3.8-flash",
    contents="Qual é a idade estimada do Universo de acordo com o CMB?",
)

# Exibe o texto retornado pelo modelo de IA
print(response.text)
```
