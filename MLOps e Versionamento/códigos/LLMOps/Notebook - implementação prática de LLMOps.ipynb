# LLMOps na Prática

## 1. Problema de negócio e requisitos
Usuários precisam consultar rapidamente regras distribuídas em um “Regulamento de Seguro Fictício”. O objetivo é responder perguntas em português com trechos recuperados do PDF, mostrar fontes e abster-se quando não houver evidência. Requisitos: ingestão do PDF, busca semântica e respostas fundamentadas.

## 2. Preparação e governança dos dados
A fonte é o PDF “Regulamento de Seguro Fictício”, Condições Gerais do Seguro Proteção Residencial Aurora, versão 1.0 (novembro de 2026). Abrange planos, coberturas, limites, franquias, carências, exclusões e procedimentos. Controle versão, integridade e autorização de uso; preserve página e origem dos trechos. Tabelas e layout podem exigir validação manual.

## 3. Seleção e configuração do modelo
O protótipo usa a modalidade gratuita do Gemini. A chave `GOOGLE_API_KEY` deve ser guardada em variável de ambiente, nunca no notebook. 

## 4. Engenharia de prompts
O prompt delimita o papel do assistente, restringe respostas ao contexto recuperado, orienta a preservar números e exceções, pedir páginas-fonte e admitir falta de evidência. Temperatura baixa ajuda a consistência, mas não elimina alucinações.

## 5. Integração RAG
Pipeline: PDF → extração por página  → contexto + pergunta enviados ao Gemini → resposta e trechos-fonte. Em produção, versionar documentos e índices, monitorar recuperação e evitar misturar versões.

## 6. Feedback

A solução pergunta se o usuário gostou ou não da resposta gerada. Essa resposta é guardada num arquivo excel para uma análise posterior.
